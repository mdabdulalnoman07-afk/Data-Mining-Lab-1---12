## Objectives
- Mine frequent itemsets using FP-Growth.
- Compare its itemsets with Apriori on the same transactions.

> Run cells from top to bottom. The notebook uses a reproducible random seed. For Adult Income labs, it attempts to load the public UCI Adult dataset and uses a clearly labelled local synthetic fallback if the download is unavailable.

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules

transactions = [
 ["bread","milk","eggs"], ["bread","butter"], ["milk","eggs"],
 ["bread","milk","butter"], ["bread","eggs"], ["milk","butter"],
 ["bread","milk","eggs","butter"], ["bread","milk"], ["eggs","butter"],
 ["bread","butter","eggs"], ["milk","bread","eggs"], ["milk","butter","bread"]
]
te = TransactionEncoder()
encoded = te.fit(transactions).transform(transactions)
basket = pd.DataFrame(encoded, columns=te.columns_)
display(basket.head())
```

## FP-Growth frequent itemsets

```python
fp_items = fpgrowth(basket, min_support=0.25, use_colnames=True)
fp_items = fp_items.sort_values("support", ascending=False)
display(fp_items)
fp_rules = association_rules(fp_items, metric="confidence", min_threshold=0.5)
display(fp_rules[["antecedents","consequents","support","confidence","lift"]].sort_values("lift", ascending=False))
```

## Compare Apriori and FP-Growth

```python
ap_items = apriori(basket, min_support=0.25, use_colnames=True)
ap_set = set(frozenset(x) for x in ap_items["itemsets"])
fp_set = set(frozenset(x) for x in fp_items["itemsets"])
print("Apriori itemsets:", len(ap_set))
print("FP-Growth itemsets:", len(fp_set))
print("Same itemsets:", ap_set == fp_set)
```

FP-Growth uses a compact FP-tree representation to avoid candidate generation in the same way as Apriori. On this small teaching dataset, runtime differences may be negligible.
